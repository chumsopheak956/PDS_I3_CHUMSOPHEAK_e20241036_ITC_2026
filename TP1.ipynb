```python
# ---------------------------------------------------------------------------
# Solution 4 Type-conversion detective
# ---------------------------------------------------------------------------
# Part A 
# Problem 4 - Part A
print(int("7") + 3.5)
print(str(10) * 3)
print(bool(""), bool(" "))
print(int(3.99))
print(float("1e3"))
print(int(True) + float("2.5"))
print(round(2.5), round(3.5))
print(type(10 / 2))
# Part B — a safe converter
def safe_int(text):
    """Turn text into an int; return None when it is not a number."""
    text = text.strip()
    if text.lstrip("-").isdigit():         # "42" or "-42"
        return int(text)
    try:
        return int(float(text))            # "3.9" -> 3
    except ValueError:
        return None                        # "abc", "", "12a"


for sample in ["42", " -8 ", "3.9", "abc", ""]:
    print(repr(sample), "->", safe_int(sample))
```

```python
# ---------------------------------------------------------------------------
# Solution 5 Letter grade
# ---------------------------------------------------------------------------
def letter_grade(score):
    """Return the letter grade of a score from 0 to 100."""
    if score < 0 or score > 100:
        return "Invalid"
    elif score >= 97:          # A+ must be tested BEFORE A
        return "A+"
    elif score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"


print()
for s in [99, 90, 85.5, 72, 60, 59.9, -5, 101]:
    print(f"{s:>5} -> {letter_grade(s)}")
```

```python
# ---------------------------------------------------------------------------
# Solution 6 Leap years and days in a month
# ---------------------------------------------------------------------------
def is_leap(year):
    """Divisible by 4, except centuries, unless divisible by 400."""
    return (year % 4 == 0 and year % 100 != 0) or year % 400 == 0


def days_in_month(month, year):
    """Return the number of days in a month (1–12), or None if invalid."""
    if month < 1 or month > 12:
        return None
    if month == 2:
        return 29 if is_leap(year) else 28
    if month in (4, 6, 9, 11):
        return 30
    return 31


print()
for y in [1900, 2000, 2024, 2026]:
    print(y, "leap" if is_leap(y) else "not leap")
print("Feb 2024:", days_in_month(2, 2024), "| Feb 2026:", days_in_month(2, 2026),
      "| Apr:", days_in_month(4, 2026), "| month 13:", days_in_month(13, 2026))

leap_count = 0
for y in range(2000, 2101):
    if is_leap(y):
        leap_count += 1
print("Leap years from 2000 to 2100:", leap_count)
```

```python
# ---------------------------------------------------------------------------
# Solution 7 Digit tools (while loops with // and %)
# ---------------------------------------------------------------------------
def digit_sum(n):
    """Add up the digits of a whole number."""
    n = abs(n)
    total = 0
    while n > 0:
        total += n % 10        # last digit
        n //= 10               # drop the last digit
    return total


def reverse_number(n):
    """Return the digits of n in reverse order, e.g. 1234 -> 4321."""
    result = 0
    while n > 0:
        result = result * 10 + n % 10
        n //= 10
    return result


def is_palindrome(n):
    """True when n reads the same in both directions."""
    return n == reverse_number(n)


print()
for n in [2026, 12321, 9875]:
    print(f"{n}: digit sum = {digit_sum(n)}, reversed = {reverse_number(n)}, "
          f"palindrome = {is_palindrome(n)}")

count = 0
for n in range(10, 1000):
    if is_palindrome(n):
        count += 1
print("Palindromes from 10 to 999:", count)
```

```python
# ---------------------------------------------------------------------------
# Solution 8 *args, **kwargs and lambda
# ---------------------------------------------------------------------------
def describe(*numbers, decimals=2, label="Data"):
    """Return (count, min, max, mean) of any number of values and print them."""
    if len(numbers) == 0:
        print(f"{label}: no values")
        return None
    mean = round(sum(numbers) / len(numbers), decimals)
    print(f"{label}: n={len(numbers)} min={min(numbers)} max={max(numbers)} mean={mean}")
    return len(numbers), min(numbers), max(numbers), mean


def make_tag(text, **style):
    """Wrap text in an HTML-like tag built from keyword arguments."""
    attrs = ""
    for key, value in style.items():
        attrs += f' {key}="{value}"'
    return f"<span{attrs}>{text}</span>"


print()
describe(12, 7, 30, 4)
describe(3.14159, 2.71828, decimals=3, label="Constants")
describe()
print(make_tag("Hello", color="red", size=14))

square = lambda x: x ** 2
is_adult = lambda age: age >= 18
print(square(9), is_adult(17), is_adult(21))
```

```python
# ---------------------------------------------------------------------------
# Solution 10 Prime numbers and Goldbach's conjecture
# ---------------------------------------------------------------------------
def is_prime(n):
    """True when n is a prime number. Tests divisors up to the square root."""
    if n < 2:
        return False
    d = 2
    while d * d <= n:
        if n % d == 0:
            return False
        d += 1
    return True


print("\nPrimes below 50:", end=" ")
count = 0
for n in range(50):
    if is_prime(n):
        print(n, end=" ")
for n in range(100):
    if is_prime(n):
        count += 1
print("\nNumber of primes below 100:", count)


def goldbach(n):
    """Return the first pair of primes (p, q) with p + q == n, or None."""
    for p in range(2, n // 2 + 1):
        if is_prime(p) and is_prime(n - p):
            return p, n - p
    return None


for n in range(4, 31, 2):
    p, q = goldbach(n)
    print(f"{n} = {p} + {q}")

# Check the conjecture for every even number from 4 to 10 000
all_ok = True
for n in range(4, 10001, 2):
    if goldbach(n) is None:
        all_ok = False
        print("Counter-example:", n)
        break
print("Goldbach holds for every even number from 4 to 10000:", all_ok)

```
